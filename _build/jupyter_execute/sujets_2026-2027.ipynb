# Liste et attribution des sujets

## Sujet 1: Équations différentielles du premier ordre 
Sujet attribué aux personnes figurant dans la première moitié de la liste d'émargement des groupes de TP A, C, E, G, I, K et LDDa.
### Objectifs :
- Démontrer que la solution d'une équation différentielle du premier ordre à coefficients constants sans second membre de la forme $\frac{\mathrm{d}x}{\mathrm{d}t} + a x = 0,\quad a \in \mathbb{R}$ s'exprime sous la forme d'une somme dont un de termes s'écrit $x(t) = C e^{-at}$, où la constante $C$ dépend d'une condition initiale ou particulière de la fonction $x$.
- Démontrer que la solution d'une équation différentielle du premier ordre à coefficients constants et second membre constant de la forme $\frac{\mathrm{d}x}{\mathrm{d}t} + a x = b,\quad a, b \in \mathbb{R}$ s'exprime sous la forme d'une somme dont un de termes s'écrit $x(t) = C e^{-at}$, où la constante $C$ dépend d'une condition initiale ou particulière de la fonction $x$.

## Sujet 2: Moment de force
Sujet attribué aux personnes figurant dans la seconde moitié de la liste d'émargement des groupes de TP A, C, E, G, I, K et LDDa.
### Objectifs :
- Une tige rigide libre de tourner autour de son point de pivot $O$ est soumis à une force $\vec{F}$ dont le point d'application est distante de $OA$ du pivot $OA$ : expliquer comment l'intensité, la direction et le sens de la force détermine le mouvement de rotation imposé par $\vec{F}$.

## Sujet 3: Cinématique en coordonnées polaires
Sujet attribué aux personnes figurant dans la première moitié de la liste d'émargement des groupes de TP B, D, F, H, J, L et LDDb.
### Objectifs :
- Expliquer les expressions des vecteurs position, vitesse et accélération exprimés en coordonnées polaires ci-dessous :
\begin{align}
\vec{OM} &= r\vec{u}_r \\
\vec{v}  &= \dot{r}\vec{u}_r + r\dot{\theta} \vec{u}_\theta \\
\vec{a}  &= (\ddot{r} - r \dot{\theta}^2) \vec{u}_r + (r\ddot{\theta} + 2\dot{r}\dot{\theta}) \vec{u}_\theta
\end{align}

## Sujet 4: Miroirs sphériques
Sujet attribué aux personnes figurant dans la seconde moitié de la liste d'émargement des groupes de TP B, D, F, H, J, L et LDDb.
### Objectifs :
- Expliquer comment se justifie la relation de conjugaison d'un miroir sphérique, notamment ses conditions de validité
- Donner une méthode de construction d'image par tracé de rayons particuliers.
- Expliquer comment la relation de conjugaison justifie le comportement des rayons particuliers.
